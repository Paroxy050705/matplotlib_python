# Bar Plot

```python
import matplotlib.pyplot as plt
x=["Python","c","c++","java"]
y=[85,70,60,82]
plt.bar(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=["Python","java","javascript","c","c++"]
y=[95,80,67,58,45]
c=["r","g","b","y","m"]
plt.xlabel("Language",fontsize=30)
plt.ylabel("Percentage",fontsize=30)
plt.title("Popularity of Languages",fontsize=45)
plt.bar(x,y,color=c,width=0.7,edgecolor="000",linewidth=5,linestyle=":",alpha=0.5,label="Graph")
plt.legend()
plt.show()
```

# Scatter Plot

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6,7,8]
y=[2,4,6,8,10,12,14,16]
size=[100,600,90,50,900,500,30,700]
color=[10,20,30,40,50,60,70,80]
plt.scatter(x,y,c=color,label="Graph",s=size,marker="*",edgecolor="000",linewidth=2)
plt.colorbar()
plt.legend()
plt.xlabel("Day",fontsize=15)
plt.ylabel("Number",fontsize=15)
plt.title("Scatter Plot",fontsize=30)
plt.show()
```

# Pie Plot

```python
import matplotlib.pyplot as plt
x=[20,30,70,40,20,5]
y=["c","c++","Python","Javascript","java","html"]
ex=[0.0,0.0,0.0,0.0,0.3,0.5]
c=["r","g","m","y","000","b"]
plt.pie(x,labels=y,explode=ex,colors=c,autopct="%0.1f%%",shadow=True,radius=1.0,textprops={"fontsize":10},wedgeprops={"linewidth":10})
plt.title("Graph",fontsize=20)

plt.show()
```
